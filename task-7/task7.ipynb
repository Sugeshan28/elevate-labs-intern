```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("data.csv")
df
```

```python
df.drop(['id'], axis=1, inplace=True)
```

```python
df
```

```python
df.replace({
    'M':0,
    'B':1
},inplace=True)
```

```python
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

x = df.drop("diagnosis", axis=1)
y = df.diagnosis

x_train,x_test,y_train,y_test = train_test_split(x,y,train_size=0.2, random_state=42)

stansclr = StandardScaler()
x_scaled = stansclr.fit_transform(x_train)
stansclr.fit_transform(x_test)


pca = PCA(n_components=2)
x_train_pca = pca.fit_transform(x_scaled)
x_test_pca = pca.transform(x_scaled)



model_li = SVC(kernel='linear', C=1.0)
model_li.fit(x_train_pca,y_train)


model_rbc = SVC(kernel='rbf', C=1.0, gamma='scale')
model_rbc.fit(x_train_pca,y_train)

y_pred = model_li.predict(x_test_pca)
y_predd = model_rbc.predict(x_test_pca)

print(x_train.shape)       # should be (113, n_features)
print(y_train.shape)       # should be (113,)
print(x_test.shape)        # should be (456, n_features)
print(y_test.shape)        # should be (456,)
print(x_train_pca.shape)   # should also be (113, 2) after PCA
print(x_test_pca.shape)

print(f"Model accuracy for kernal {accuracy_score(y_train,y_pred)}")

print(f"Model accuracy for rbf {accuracy_score(y_train,y_predd)}")
```

```python
def plot_decision_boundary(model, X, y, title):
    # Convert DataFrame to NumPy array if needed
    if hasattr(X, "values"):
        X = X.values
    if hasattr(y, "values"):
        y = y.values

    h = 0.02
    x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    xx, yy = np.meshgrid(np.arange(x_min, x_max, h),
                         np.arange(y_min, y_max, h))

    Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)

    plt.contourf(xx, yy, Z, alpha=0.3, cmap=plt.cm.coolwarm)
    plt.scatter(X[:, 0], X[:, 1], c=y, edgecolors='k', cmap=plt.cm.coolwarm)
    plt.title(title)
    plt.xlabel("Feature 1")
    plt.ylabel("Feature 2")
    


plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plot_decision_boundary(model_li, x_train, y_train, "SVM with Linear Kernel")

plt.subplot(1, 2, 2)
plot_decision_boundary(model_rbc, x_train, y_train, "SVM with RBF Kernel")

plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import cross_val_score

scores_linear = cross_val_score(model_li, x, y, cv=5)
scores_rbf = cross_val_score(model_rbc, x, y, cv=5)

# Report average accuracy
print(f"Linear SVM Cross-Validation Accuracy: {scores_linear.mean():.4f}")
print(f"RBF SVM Cross-Validation Accuracy: {scores_rbf.mean():.4f}")
```
