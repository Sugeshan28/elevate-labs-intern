```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("winequality-red.csv")
df
```

```python
df.describe()
```

Histogram

```python
df_columns = df.select_dtypes(include='number').columns

df[df_columns].hist(bins = 10,figsize=(20,15), layout=(6,6), edgecolor = 'black')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
box_col = df.select_dtypes(include='number').columns

fig,axes = plt.subplots(3,4, figsize=(10,15))

axes = axes.flatten()


for i,cols in enumerate(box_col):

    df.boxplot(column=cols, ax=axes[i])
    axes[i].set_title(cols)

plt.tight_layout()
plt.show()
```

```python
import seaborn as sns

sns.pairplot(df,hue='pH')
```

```python
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.show()
```

```python
from scipy.stats import zscore

z_score = df.select_dtypes(include='number').apply(zscore)
out = (z_score.abs() > 3).sum()
print(out)
```
